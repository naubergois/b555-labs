# Keras exercício 10 — Functional API — dois ramos densos

**B555** · Área: `api` · Dataset: sklearn load_digits (split de features)

**O que você aprende:** Functional API quando Sequential não basta: ramos, concat, multi-entrada.

**Métrica:** accuracy no teste

**Limite:** Grafo complexo sem necessidade vira dívida técnica.

Frase-guia: *a rede calcula; nós decidimos*.


```python
%pip install -q tensorflow scikit-learn numpy
import numpy as np
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
tf.random.set_seed(7); np.random.seed(7)
X, y = load_digits(return_X_y=True)
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.25, random_state=7, stratify=y)
sc = StandardScaler(); Xtr, Xte = sc.fit_transform(Xtr), sc.transform(Xte)
inp = keras.Input(shape=(64,))
a = layers.Dense(32, activation='relu')(inp)
b = layers.Dense(32, activation='relu')(inp)
x = layers.Concatenate()([a, b])
out = layers.Dense(10, activation='softmax')(x)
model = keras.Model(inp, out)
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.fit(Xtr, ytr, validation_split=0.2, epochs=25, verbose=0)
print('acc teste', model.evaluate(Xte, yte, verbose=0)[1])
print('Functional API: use quando Sequential não basta.')
```
