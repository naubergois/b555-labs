# Keras exercício 07 — Desbalanceamento — class_weight

**B555** · Área: `metricas` · Dataset: sklearn make_classification (weights desbalanceados)

**O que você aprende:** Classes raras: accuracy mente. class_weight e recall da classe crítica.

**Métrica:** recall da classe minoritária

**Limite:** Peso alto demais explode falso positivo.

Frase-guia: *a rede calcula; nós decidimos*.


```python
%pip install -q tensorflow scikit-learn numpy
import numpy as np
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
from sklearn.utils.class_weight import compute_class_weight
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
tf.random.set_seed(7); np.random.seed(7)
X, y = make_classification(n_samples=3000, weights=[0.9, 0.1], n_features=20, random_state=7)
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.25, random_state=7, stratify=y)
cw = compute_class_weight('balanced', classes=np.unique(ytr), y=ytr)
class_weight = {i: float(w) for i, w in enumerate(cw)}
print('class_weight', class_weight)
model = keras.Sequential([layers.Input(shape=(20,)), layers.Dense(32, activation='relu'), layers.Dense(1, activation='sigmoid')])
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy', keras.metrics.Recall(name='recall')])
model.fit(Xtr, ytr, validation_split=0.2, epochs=30, class_weight=class_weight, verbose=0)
pred = (model.predict(Xte, verbose=0).ravel() >= 0.5).astype(int)
print(classification_report(yte, pred, digits=3))
print('LIMITE: weight alto demais aumenta falso positivo.')
```
