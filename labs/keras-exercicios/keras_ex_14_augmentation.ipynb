# Keras exercício 14 — Data augmentation — RandomFlip/Rotation

**B555** · Área: `visao` · Dataset: keras.datasets.fashion_mnist (subset)

**O que você aprende:** Aumentar diversidade sintética reduz overfit quando há poucas imagens.

**Métrica:** val_accuracy com/sem aug

**Limite:** Augmentation absurda inventa exemplos impossíveis.

Frase-guia: *a rede calcula; nós decidimos*.


```python
%pip install -q tensorflow numpy
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
tf.random.set_seed(7); np.random.seed(7)
(xtr, ytr), (xte, yte) = keras.datasets.fashion_mnist.load_data()
# subset para sentir o efeito do aug
xtr, ytr = xtr[:8000], ytr[:8000]
xtr = xtr.astype('float32')[..., None] / 255.0
xte = xte.astype('float32')[..., None] / 255.0
aug = keras.Sequential([layers.RandomFlip('horizontal'), layers.RandomRotation(0.05)])
def build():
    inp = keras.Input(shape=(28, 28, 1))
    x = aug(inp)
    x = layers.Conv2D(32, 3, activation='relu')(x)
    x = layers.MaxPooling2D()(x)
    x = layers.Flatten()(x)
    out = layers.Dense(10, activation='softmax')(x)
    m = keras.Model(inp, out)
    m.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
    return m
model = build()
model.fit(xtr, ytr, validation_split=0.15, epochs=4, batch_size=128, verbose=1)
print('acc teste', model.evaluate(xte, yte, verbose=0)[1])
```
