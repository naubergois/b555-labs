# Keras exercício 20 — Portfolio — escolha 1 caso + briefing ético

**B555** · Área: `portfolio` · Dataset: à escolha (Iris / Fashion / IMDB / série)

**O que você aprende:** Fechar o ciclo: escolher domínio, métrica, limite escrito e próxima ação humana.

**Métrica:** briefing de 5 linhas + métrica de teste

**Limite:** Portfolio sem limite ético é só demo.

Frase-guia: *a rede calcula; nós decidimos*.


```python
%pip install -q tensorflow scikit-learn numpy
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
tf.random.set_seed(7); np.random.seed(7)
# Exemplo mínimo — troque pelo caso que você escolheu (Fashion/IMDB/série)
X, y = load_iris(return_X_y=True)
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.25, random_state=7, stratify=y)
sc = StandardScaler(); Xtr, Xte = sc.fit_transform(Xtr), sc.transform(Xte)
model = keras.Sequential([layers.Input(shape=(4,)), layers.Dense(16, activation='relu'), layers.Dense(3, activation='softmax')])
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.fit(Xtr, ytr, epochs=30, verbose=0)
acc = accuracy_score(yte, model.predict(Xte, verbose=0).argmax(1))
print('Métrica teste (accuracy exemplo):', round(float(acc), 4))
briefing = '''
1. Pergunta: ...
2. Dado: ...
3. Métrica e por quê: ...
4. Limite do modelo: ...
5. Próxima ação humana: ...
'''
print(briefing)
print('Preencha o briefing de verdade. Sem limite ético, é só demo.')
print('A rede calcula; nós decidimos.')
```
