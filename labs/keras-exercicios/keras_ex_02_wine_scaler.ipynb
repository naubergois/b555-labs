# Keras exercício 02 — Wine — StandardScaler só no treino

**B555** · Área: `tabular` · Dataset: sklearn load_wine

**O que você aprende:** Vazamento de escala: fit do scaler no teste infla métrica. Fit só no treino; transform no teste.

**Métrica:** accuracy no teste com e sem vazamento (demonstração)

**Limite:** Scaler correto não garante boa decisão de negócio.

Frase-guia: *a rede calcula; nós decidimos*.


```python
%pip install -q tensorflow scikit-learn numpy
import numpy as np
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
tf.random.set_seed(7); np.random.seed(7)
X, y = load_wine(return_X_y=True)
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.25, random_state=7, stratify=y)
# CORRETO: fit só no treino
sc = StandardScaler(); Xtr_s, Xte_s = sc.fit_transform(Xtr), sc.transform(Xte)
model = keras.Sequential([layers.Input(shape=(Xtr.shape[1],)), layers.Dense(32, activation='relu'), layers.Dense(3, activation='softmax')])
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.fit(Xtr_s, ytr, epochs=40, verbose=0)
print('acc teste (scaler honesto)', model.evaluate(Xte_s, yte, verbose=0)[1])
print('LIMITE: scaler certo ≠ decisão de negócio.')
```
