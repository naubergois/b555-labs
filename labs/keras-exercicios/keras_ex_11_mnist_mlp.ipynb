# Keras exercício 11 — MNIST — MLP em pixels achatados

**B555** · Área: `visao` · Dataset: keras.datasets.mnist

**O que você aprende:** Imagem 28×28 vira vetor 784. MLP funciona, mas não respeita localidade espacial.

**Métrica:** accuracy no teste

**Limite:** MLP em pixel não é CNN — sofre em deslocamento/rotação.

Frase-guia: *a rede calcula; nós decidimos*.


```python
%pip install -q tensorflow numpy
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
tf.random.set_seed(7); np.random.seed(7)
(xtr, ytr), (xte, yte) = keras.datasets.mnist.load_data()
xtr = xtr.reshape(-1, 784).astype('float32') / 255.0
xte = xte.reshape(-1, 784).astype('float32') / 255.0
model = keras.Sequential([layers.Input(shape=(784,)), layers.Dense(128, activation='relu'), layers.Dropout(0.2), layers.Dense(10, activation='softmax')])
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.fit(xtr, ytr, validation_split=0.1, epochs=5, batch_size=128, verbose=1)
print('acc teste', model.evaluate(xte, yte, verbose=0)[1])
print('LIMITE: MLP em pixel ignora geometria local.')
```
