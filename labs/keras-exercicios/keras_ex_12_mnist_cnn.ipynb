**O que você aprende:** Convolução captura padrões locais; pooling reduz resolução.

**Métrica:** accuracy no teste

**Não é dispositivo médico.** Apoio didático apenas.


```python
%pip install -q tensorflow numpy
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
tf.random.set_seed(7); np.random.seed(7)
(xtr, ytr), (xte, yte) = keras.datasets.mnist.load_data()
xtr = xtr.astype('float32')[..., None] / 255.0
xte = xte.astype('float32')[..., None] / 255.0
model = keras.Sequential([
    layers.Input(shape=(28, 28, 1)),
    layers.Conv2D(32, 3, activation='relu'), layers.MaxPooling2D(),
    layers.Conv2D(64, 3, activation='relu'), layers.MaxPooling2D(),
    layers.Flatten(), layers.Dense(64, activation='relu'), layers.Dense(10, activation='softmax'),
])
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.fit(xtr, ytr, validation_split=0.1, epochs=3, batch_size=128, verbose=1)
print('acc teste', model.evaluate(xte, yte, verbose=0)[1])
print('LIMITE: MNIST ≠ imagem clínica.')
```
