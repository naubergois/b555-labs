# Keras exercício 19 — Autoencoder — reconstrução MNIST

**B555** · Área: `nao_supervisionado` · Dataset: keras.datasets.mnist

**O que você aprende:** Encoder comprime; decoder reconstrói. Erro alto pode sinalizar anomalia didática.

**Métrica:** MSE de reconstrução

**Limite:** Erro de reconstrução ≠ diagnóstico de fraude real sem calibração.

Frase-guia: *a rede calcula; nós decidimos*.


```python
%pip install -q tensorflow numpy
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
tf.random.set_seed(7); np.random.seed(7)
(xtr, _), (xte, _) = keras.datasets.mnist.load_data()
xtr = xtr.astype('float32').reshape(-1, 784) / 255.0
xte = xte.astype('float32').reshape(-1, 784) / 255.0
enc_in = keras.Input(shape=(784,))
z = layers.Dense(64, activation='relu')(enc_in)
z = layers.Dense(16, activation='relu')(z)
dec = layers.Dense(64, activation='relu')(z)
out = layers.Dense(784, activation='sigmoid')(dec)
ae = keras.Model(enc_in, out)
ae.compile(optimizer='adam', loss='mse')
ae.fit(xtr, xtr, validation_split=0.1, epochs=5, batch_size=256, verbose=1)
rec = ae.predict(xte[:1000], verbose=0)
mse = float(np.mean((xte[:1000] - rec) ** 2))
print('MSE reconstrução', round(mse, 5))
print('LIMITE: erro alto ≠ fraude real sem calibração.')
```
