# Keras exercício 05 — BatchNormalization entre densas

**B555** · Área: `regularizacao` · Dataset: sklearn make_classification

**O que você aprende:** BN estabiliza a distribuição das ativações e pode acelerar o treino.

**Métrica:** val_accuracy

**Limite:** BN muda comportamento em batch=1 / inferência — leia a doc.

Frase-guia: *a rede calcula; nós decidimos*.


```python
%pip install -q tensorflow scikit-learn numpy
import numpy as np
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
tf.random.set_seed(7); np.random.seed(7)
X, y = make_classification(n_samples=2500, n_features=30, random_state=7)
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.25, random_state=7)
model = keras.Sequential([
    layers.Input(shape=(30,)),
    layers.Dense(64), layers.BatchNormalization(), layers.Activation('relu'),
    layers.Dense(32), layers.BatchNormalization(), layers.Activation('relu'),
    layers.Dense(1, activation='sigmoid'),
])
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model.fit(Xtr, ytr, validation_split=0.2, epochs=25, verbose=0)
print('acc teste', model.evaluate(Xte, yte, verbose=0)[1])
```
