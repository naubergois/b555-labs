# Keras exercício 04 — A/B Dropout — overfit visível

**B555** · Área: `regularizacao` · Dataset: sklearn make_classification

**O que você aprende:** Dropout desliga unidades no treino e reduz memorização. Compare curvas com/sem dropout.

**Métrica:** val_loss e gap treino–val

**Limite:** Dropout demais subajusta; não é mágica.

Frase-guia: *a rede calcula; nós decidimos*.


```python
%pip install -q tensorflow scikit-learn numpy
import numpy as np
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
tf.random.set_seed(7); np.random.seed(7)
X, y = make_classification(n_samples=2000, n_features=20, n_informative=8, random_state=7)
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.25, random_state=7)
def build(drop):
    m = keras.Sequential([layers.Input(shape=(20,)), layers.Dense(128, activation='relu'), layers.Dropout(drop), layers.Dense(64, activation='relu'), layers.Dropout(drop), layers.Dense(1, activation='sigmoid')])
    m.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
    return m
h0 = build(0.0).fit(Xtr, ytr, validation_split=0.2, epochs=30, verbose=0)
h1 = build(0.4).fit(Xtr, ytr, validation_split=0.2, epochs=30, verbose=0)
print('sem dropout val_acc', h0.history['val_accuracy'][-1])
print('com dropout val_acc', h1.history['val_accuracy'][-1])
print('Olhe o gap treino–val: dropout costuma reduzir memorização.')
```
