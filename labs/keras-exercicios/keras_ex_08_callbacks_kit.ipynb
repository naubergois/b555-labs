**O que você aprende:** Parar cedo e salvar o melhor peso em disco. restore_best_weights.

**Métrica:** val_accuracy + arquivo .keras

**Limite:** Monitor errado (ex.: loss quando importa AUC) para no lugar errado.

Frase-guia: *a rede calcula; nós decidimos*.


```python
%pip install -q tensorflow scikit-learn numpy
import numpy as np
from pathlib import Path
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
tf.random.set_seed(7); np.random.seed(7)
X, y = load_digits(return_X_y=True)
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.25, random_state=7, stratify=y)
sc = StandardScaler(); Xtr, Xte = sc.fit_transform(Xtr), sc.transform(Xte)
ckpt = Path('/tmp/keras_ex08_best.keras')
model = keras.Sequential([layers.Input(shape=(64,)), layers.Dense(64, activation='relu'), layers.Dense(10, activation='softmax')])
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
cb = [
    keras.callbacks.EarlyStopping(monitor='val_accuracy', patience=5, restore_best_weights=True),
    keras.callbacks.ModelCheckpoint(str(ckpt), monitor='val_accuracy', save_best_only=True),
]
model.fit(Xtr, ytr, validation_split=0.2, epochs=40, callbacks=cb, verbose=0)
print('acc teste', model.evaluate(Xte, yte, verbose=0)[1], 'ckpt', ckpt.exists())
```
