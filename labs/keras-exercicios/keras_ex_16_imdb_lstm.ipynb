# Keras exercício 16 — IMDB — LSTM unária

**B555** · Área: `texto` · Dataset: keras.datasets.imdb

**O que você aprende:** LSTM lê sequência; ordem das palavras importa mais que bag-of-words.

**Métrica:** accuracy no teste

**Limite:** LSTM rasa não entende ironia nem contexto longo.

Frase-guia: *a rede calcula; nós decidimos*.


```python
%pip install -q tensorflow numpy
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
tf.random.set_seed(7); np.random.seed(7)
vocab = 8000
(xtr, ytr), (xte, yte) = keras.datasets.imdb.load_data(num_words=vocab)
xtr = keras.preprocessing.sequence.pad_sequences(xtr, maxlen=150)
xte = keras.preprocessing.sequence.pad_sequences(xte, maxlen=150)
model = keras.Sequential([
    layers.Input(shape=(150,)),
    layers.Embedding(vocab, 32),
    layers.LSTM(32),
    layers.Dense(1, activation='sigmoid'),
])
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model.fit(xtr, ytr, validation_split=0.1, epochs=2, batch_size=128, verbose=1)
print('acc teste', model.evaluate(xte, yte, verbose=0)[1])
```
