```python
# Célula 1 — deps + clone do repositório (código completo dos 100+ exercícios)
%pip install -q numpy pandas scikit-learn matplotlib

import os
import sys
import subprocess
from pathlib import Path

REPO_URL = "https://github.com/naubergois/redesneurais.git"
REPO_NAME = "redesneurais"
base = Path("/content") if Path("/content").is_dir() else Path.cwd()
repo = base / REPO_NAME

marker = repo / "labs" / "exercicios" / "common.py"
if not marker.is_file():
    # Se já estamos DENTRO do repo clonado/aberto pelo Colab a partir do GitHub:
    here = Path.cwd()
    if (here / "labs" / "exercicios" / "common.py").is_file():
        repo = here
    elif (here.parent / "labs" / "exercicios" / "common.py").is_file():
        repo = here.parent
    else:
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, str(repo)], check=True)

os.chdir(repo)
sys.path.insert(0, str(repo / "labs" / "exercicios"))
print("OK · cwd =", Path.cwd())
print("script =", "labs/exercicios/ex_052.py")
assert Path("labs/exercicios/ex_052.py").is_file(), "Script não encontrado — faça push do repo no GitHub"
```
