# Keras exercício 18 — Série sintética — LSTM

**B555** · Área: `series` · Dataset: seno + ruído sintético

**O que você aprende:** LSTM guarda estado ao longo da janela; compare MAE com o baseline Dense.

**Métrica:** MAE one-step vs Dense

**Limite:** LSTM ≠ previsão garantida de demanda real.

Frase-guia: *a rede calcula; nós decidimos*.


```python
%pip install -q tensorflow numpy
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
tf.random.set_seed(7); np.random.seed(7)
t = np.linspace(0, 40, 2000)
serie = np.sin(t) + 0.1 * np.random.randn(len(t))
win = 20
Xs, ys = [], []
for i in range(len(serie) - win):
    Xs.append(serie[i:i+win]); ys.append(serie[i+win])
Xs, ys = np.array(Xs)[..., None], np.array(ys)
n = int(0.8 * len(Xs)); Xtr, Xte, ytr, yte = Xs[:n], Xs[n:], ys[:n], ys[n:]
model = keras.Sequential([layers.Input(shape=(win, 1)), layers.LSTM(32), layers.Dense(1)])
model.compile(optimizer='adam', loss='mse', metrics=['mae'])
model.fit(Xtr, ytr, validation_split=0.1, epochs=8, batch_size=64, verbose=0)
mae = model.evaluate(Xte, yte, verbose=0)[1]
print('MAE one-step (LSTM)', round(float(mae), 4))
print('LIMITE: série sintética ≠ demanda real.')
```
